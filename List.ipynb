### zip function
zip() takes two or more lists (or any iterables) and combines them element by element into tuples.

Each tuple contains one element from each list at the same index.

```python
names = ['Alice', 'Bob', 'Charlie']
scores = [85, 90, 95]

zipped = zip(names, scores)
print(list(zipped))
```

```python
print(list(zipped)) # only executed at once
```

##### zip the list tuple and string and return the tuple

```python
a = [1, 2, 3]
b = ('a', 'b', 'c')
c = 'xyz'

z = zip(a, b, c)
print(list(z))  # [(1, 'a', 'x'), (2, 'b', 'y'), (3, 'c', 'z')]
```

#### we get the keys of dict

```python
a = {"x": 1, "y": 2}
b = {"x": 100, "z": 200}

print(list(zip(a, b)))  # [('x', 'x'), ('y', 'z')]
```

```python
print(tuple(zip(a.values(), b.values())))
```
