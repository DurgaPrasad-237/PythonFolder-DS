## finding address

```python
x = "hello";
y = "hello";
```

### see address will be same for both x and y

```python
print(f"x : address{id(x)}")
print(f"y : address{id(y)}")
```

```python
list1 = [1,2,3];
list2 = [1,2,3];
```

### now address would be different

```python
print(f"list1 : address{id(list1)}")
print(f"list2 : address{id(list2)}")
```

### lets see the tuples

```python
tuple1 = (1,2,3);
tuple2 = (1,2,3);
```

### tuple1 and tuple2 can be different even though they are identical. This happens because Python doesn’t always reuse memory for immutable objects (like tuples), especially when created separately in a non-interactive context.

```python
print(f"tuple1 : address{id(tuple1)}")
print(f"tuple2 : address{id(tuple2)}")
```

## **is keyword** (it compare the identity of the two objects check checks whether two variables refer to the same object in memory.)

```python
str1 = "lahari";
str2 = "lahari";
```

```python
str1 is str2
```

```python
tuple1 is tuple2
```

```python
list1 is list2
```

## The **in keyword** is used to check if an element exists within an iterable 

```python
"h" in str1
```

```python
"z" in str1
```

```python
3 in list1
```

```python
4 in list1
```
