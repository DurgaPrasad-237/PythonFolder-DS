### if else conditional statements

#### check positive or negative

```python
number = int(input());
if number > 0:
    print("Positive");
elif number < 0:
    print("Negative");
else:
    print("Zero")
```

#### find the largest of two numbers

```python
number1 = int(input());
number2 = int(input());
if number1 > number2:
    print(f"{number1} is greater than {number2}")
elif number2 > number1:
     print(f"{number2} is greater than {number1}")
else:
     print(f"{number2} and {number1} are equal")
```

#### number is even or odd

```python
number = int(input());
if number % 2 == 0:
    print("Even Number")
else:
    print("Odd Number")
```

#### leap year

```python
year = int(input())
if year % 4 == 0 and year % 100 == 0:
    if year % 400:
        print("Leap Year")
    else:
         print("Not Leap Year")
elif year % 4 == 0:
    print("Leap Year")
else:
    print("Not Leap Year")
```

#### check character vowel or consonant

```python
character = input();
vowels = 'AEIOUaeiou'
if character in vowels:
    print("Vowel")
else:
    print("Consonant")
```

#### assign a grade based on marks

```python
marks = int(input())

match marks:
    case m if 90 < m <= 100: 
        print("A Grade")
    case m if 80 < m < 90: 
        print("B Grade")
    case m if 70 < m < 80:
        print("C Grade")
    case _:
        print("fail")
```
