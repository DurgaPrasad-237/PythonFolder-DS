### enumerate

```python
colors = ['red', 'blue', 'green']
for i, color in enumerate(colors, start=10):
    print(i, color)
```

### zip

zip() takes two or more lists (or any iterables) and combines them element by element into tuples.

Each tuple contains one element from each list at the same index.

```python
names = ['Alice', 'Bob', 'Charlie']
scores = [85, 90, 95]

zipped = zip(names, scores)
print(list(zipped))
```

## Real Life Example using List

```python
print("1.Add new item\n2.Sort items by alphabetically\n3.addTocart\n4.Display items left")
items_list = ["milk","bread","eggs","curd"]
status = True
cart = []
def addItems():
    count = int(input("How many items you want to add"))
    for i in range(count):
        items_list.append(input(f"Enter the {i+1} item:"))
    print(items_list)

def sortItems():
    items_list.sort()
    print(items_list)

def addtoCart():
    buystatus = True
    while(buystatus):
        item = input(f"Enter the item you want here the list:\n{items_list}").lower()
        if (item not in items_list):
            print(f"You enter the wrong item")
        else:
            cart.append(item)
            items_list.remove(item)
            buystatus =  True if (input("Do you wish to add one more yes/no").lower()) == "yes" else False
    print(f"Thank you visit again here is your list:\n{cart}")

def displayItemsLeft():
    print(f"here left items:{items_list}")
    
while(status):
    try:
        check = int(input("Enter number you want to check (1–4): "))
        if check not in [1, 2, 3, 4]:
            print("Invalid choice. Please enter a number between 1 and 4.")
            continue
    except ValueError:
        print("Invalid input. Please enter a valid number.")
        continue
    match check:
        case 1:
            addItems()
        case 2:
            sortItems()
        case 3:
            addtoCart()
        case 4:
            displayItemsLeft()
    status = True if (input("Do you want to continue yes/no").lower()) == "yes" else False
```
