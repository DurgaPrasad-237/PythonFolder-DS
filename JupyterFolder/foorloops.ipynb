### loops

#### forloop

pattern

```python
row =  int(input())
for x in range(row + 1):
    for j in range(x):
        print('*', end="")
    print()
```

sum of even numbers

```python
num = int(input())
count = 0;
for i in range(num + 1):
    if i % 2 == 0:
        count += i
print(count);
```

reverse a number

```python
num = input()
print(num[::-1])
```

```python
num = input()
for i in range(len(num) -1,-1,-1):
    print(num[i])
```

#### while loop

multiplication

```python
num = int(input());
i = 1;
while(i <= 10):
    print(f"{num} * {i} = {num * i}")
    i += 1
```

factorial

```python
num = int(input())
fact = 1;
while(num > 0):
    fact *= num
    num -= 1
print(fact)
```

```python
print(*[i for i in range(5)])
# The * operator is used to unpack the list. It takes each item of the list and passes it as separate arguments.
```
